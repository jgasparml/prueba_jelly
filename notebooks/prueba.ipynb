```python
import pandas as pd
df = pd.read_csv('prueba.csv')
# Limpieza de datos: convertir columnas a numéricas y eliminar comas
numeric_cols = ['spend_usd', 'impressions', 'clicks', 'platform_conversions', 'tracked_revenue_usd']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(',', '', regex=False), errors='coerce')

# Filtro: campañas con min 1 conversión 
df_filtered = df[df['platform_conversions'] >= 1]

# Agrupación: por canal y suma de métricas numéricas
df_grouped = df_filtered.groupby('channel')[numeric_cols].sum().reset_index()

# Cálculo: ROAS y CPA
df_grouped['ROAS'] = df_grouped['tracked_revenue_usd'] / df_grouped['spend_usd']
df_grouped['CPA'] = df_grouped['spend_usd'] / df_grouped['platform_conversions']

# Renombrar: columnas para reflejar métricas totales
df_grouped = df_grouped.rename(columns={
    'spend_usd': 'spend_total',
    'clicks': 'clicks_total',
    'platform_conversions': 'conversions_total',
    'tracked_revenue_usd': 'revenue_total'
})

# Selección y Ordenamiento: columnas finales y orden por ROAS descendente
output_cols = ['channel', 'spend_total', 'clicks_total', 'conversions_total', 'revenue_total', 'ROAS', 'CPA']
df_global = df_grouped[output_cols].sort_values(by='ROAS', ascending=False)

# Visualización
df_global

#Creación: archivo CSV con métricas globales
# df_global.to_csv('global_metrics.csv', index=False)
```
